```python
jd = pd.read_csv(f'../StatisticalTestResults/POST-MYTHS/KW-T2-delta-JustDetention.csv')
ss = pd.read_csv(f'../StatisticalTestResults/POST-MYTHS/KW-T2-delta-SurvivorStories.csv')
```
