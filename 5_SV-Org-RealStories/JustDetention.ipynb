```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
base_url = "https://justdetention.org/story/"
listing_urls = [base_url] + [f"{base_url}?pageno={i}" for i in range(1, 19)]
listing_urls
```

```python
all_stories = []

for url in listing_urls:
    print(f"Scraping {url}...")
    resp = requests.get(url)
    if resp.status_code != 200:
        print(f"Failed to fetch {url}")
        continue

    soup = BeautifulSoup(resp.text, "html.parser")

    # Find all <li> elements (stories)
    for li in soup.find_all("li"):
        a_tag = li.select_one("div.story-author a")
        blurb_tag = li.select_one("div.story-views p")

        if a_tag and blurb_tag:
            story_url = a_tag.get("href")
            blurb = blurb_tag.get_text(strip=True)
            all_stories.append({"story_url": story_url, "blurb": blurb})
```

```python
df = pd.DataFrame(all_stories)
```
