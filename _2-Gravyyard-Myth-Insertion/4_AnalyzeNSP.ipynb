```python
import pandas as pd
import numpy as np
import sys
```

```python
gemini = pd.read_csv('Results/Gemini_candidates_NSP.csv')
```

```python
num_narratives = range(100)
myth_types = ['clothing', 'victim_intoxication', 'perpetrator_intoxication', 'resistance']
myth_variations = ['NegMyth', 'NegNonMyth', 'PosMyth', 'PosNonMyth']
doses = [1, 2]
```

```python
narrative_idx = num_narratives[0]
myth_type = myth_types[0]
myth_variation = myth_variations[0]
dose = doses[0]

narrative_subset = gemini[(gemini['narrative_idx'] == narrative_idx) & (gemini['myth_type'] == myth_type) & 
                            (gemini['myth_variation'] == myth_variation) & (gemini['dose'] == dose)]
```
