```python
import pandas as pd

for model in ["llama", "qwen", "mistral"]:
    print(f"\n{'='*50}")
    print(f"MODEL: {model}")
    
    df = pd.read_csv(f"Results/Advice/{model}_advice.csv", on_bad_lines='skip')
    print(f"Rows: {len(df)}")
    print(f"\nfinish_reason:\n{df['finish_reason'].value_counts()}")
    print(f"\nrefused: {df['refused'].sum()} / {len(df)} ({100*df['refused'].mean():.1f}%)")
    print(f"\nn_tokens: mean={df['n_tokens'].mean():.0f}, min={df['n_tokens'].min()}, max={df['n_tokens'].max()}")
    
    # Refusal rate by condition
    print(f"\nRefusal by condition:")
    print(df.groupby("condition")["refused"].mean().round(3))
    
    # Top refusal combinations
    myth_df = df[df["condition"] == "myth"]
    print(f"\nTop refusal rates (myth condition):")
    print(
        myth_df.groupby(["myth_type", "frame", "dose"])["refused"]
        .mean().sort_values(ascending=False).head(10).round(3)
    )
    
    # Spot check one refused response
    refused_ex = df[df["refused"]]["response"].iloc[0] if df["refused"].any() else "none"
    print(f"\nSample refused response:\n{refused_ex[:300]}")
```
